```python
import pandas as pd

df_housing = pd.read_csv("/Users/songsooyeoun/Desktop/git_test/streamlit/seoul_housing/data/seoul_real_estate_en.csv")
df_bdong = pd.read_csv("/Users/songsooyeoun/Desktop/git_test/streamlit/seoul_housing/data/bdong_eng.txt", 
                        sep="|", header=None, names=["CODE", "STDG_NM_KR", "STDG_NM_EN"], encoding="cp949")


df_housing['CODE'] = df_housing['CGG_CD'].astype(str) + df_housing['STDG_CD'].astype(str)
df_bdong['CODE'] = df_bdong['CODE'].astype(str)

df_housing = df_housing.merge(df_bdong[['CODE', 'STDG_NM_EN']], on='CODE', how='left')
df_housing['STDG_NM'] = df_housing['STDG_NM_EN']
df_housing.drop(columns=['STDG_NM_EN'], inplace=True)
display(df_housing[['CGG_CD', 'CGG_NM', 'STDG_NM']].head(10))

output_path = "/Users/songsooyeoun/Desktop/git_test/streamlit/seoul_housing/data/seoul_housing_en_updated.csv"
df_housing.to_csv(output_path, index=False, encoding="utf-8")
```

```python
import json

file_path = '/Users/songsooyeoun/Desktop/git_test/streamlit/seoul_housing/in_en.json'
df_housing = pd.read_csv("/Users/songsooyeoun/Desktop/git_test/streamlit/seoul_housing/data/seoul_housing_en_updated.csv")

with open(file_path, 'r', encoding='utf-8') as file:
    mappings = json.load(file)

# Step 2: Access the dictionaries for BLDG_USG and CGG_NM mappings
cgn_nm_mapping = mappings['cgn_nm_mapping']
lotno_se_nm_mapping = mappings['lotno_se_nm_mapping']
rght_se_mapping = mappings['rght_se_mapping']
bldg_usg_mapping = mappings['bldg_usg_mapping']
dclr_se_mapping = mappings['dclr_se_mapping']

df_housing['CGG_NM'] = df_housing['CGG_NM'].replace(cgn_nm_mapping)
df_housing['LOTNO_SE_NM'] = df_housing['LOTNO_SE_NM'].replace(lotno_se_nm_mapping)
df_housing['RGHT_SE'] = df_housing['RGHT_SE'].replace(rght_se_mapping)
df_housing['BLDG_USG'] = df_housing['BLDG_USG'].replace(bldg_usg_mapping)
df_housing['DCLR_SE'] = df_housing['DCLR_SE'].replace(dclr_se_mapping)

df_housing[['CGG_CD', 'CGG_NM', 'LOTNO_SE_NM', 'RGHT_SE', 'STDG_NM', 'BLDG_USG']].head()
df_housing.to_csv(output_path, index=False, encoding="utf-8")
```

```python
from googletrans import Translator
from tqdm import tqdm
import pandas as pd

translator = Translator()

def translate_building_name(name):
    try:
        return translator.translate(name, src='ko', dest='en').text
    except Exception as e:
        return name
        
tqdm.pandas()

df_housing['BLDG_NM_EN'] = df_housing['BLDG_NM'].progress_apply(translate_building_name)
df_housing.to_csv(output_path, index=False, encoding="utf-8")

display(df_housing[['BLDG_NM', 'BLDG_NM_EN']])
```

```python
def needs_translation(text):
    return pd.isna(text) or all('\uac00' <= char <= '\ud7a3' for char in str(text))  # 한글만 포함된 경우

mask = df_housing['BLDG_NM_EN'].apply(needs_translation)
df_housing.loc[mask, 'BLDG_NM_EN'] = df_housing.loc[mask, 'BLDG_NM'].progress_apply(translate_building_name)

df_housing.to_csv(output_path, index=False, encoding="utf-8")

display(df_housing[['BLDG_NM', 'BLDG_NM_EN']])
```
